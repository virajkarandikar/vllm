# Basic vLLM API

```python
from vllm import LLM, SamplingParams

# example of basic API usage
llm = LLM(model="TinyLlama/TinyLlama-1.1B-Chat-v1.0", max_model_len=256, gpu_memory_utilization=0.8)
sampling_params = SamplingParams(temperature=0.7, top_p=0.95, max_tokens=50)
outputs = llm.generate("My name is", sampling_params)
print(outputs[0].outputs[0].text)
```

# Async vLLM API example

```python
# example of llm streaming usage
from vllm import SamplingParams
from vllm.engine.arg_utils import AsyncEngineArgs
from vllm.v1.engine.async_llm import AsyncLLM

import torch
import matplotlib.pyplot as plt

engine_args = AsyncEngineArgs(
    model="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    max_model_len=256,
    gpu_memory_utilization=0.8,
)
engine = AsyncLLM.from_engine_args(engine_args)

sampling_params = SamplingParams(
    max_tokens=20,
    temperature=0.8,
    top_p=0.95,
    seed=42,  # For reproducible results
)

all_states_lst = []
async for output in engine.generate(request_id="1",prompt="My name is", sampling_params=sampling_params):
    all_states_lst.append(output.outputs[0].hidden_states)


all_states = torch.cat(all_states_lst, dim=0)
plt.imshow(all_states.cpu().float().numpy().T, aspect='auto')
plt.colorbar()
plt.show()
```
